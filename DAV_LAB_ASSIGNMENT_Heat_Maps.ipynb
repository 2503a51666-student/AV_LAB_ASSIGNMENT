```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Excel file
df = pd.read_excel('DAV_HEAT MAP GRAPHS.xlsx', sheet_name='Borrow_Transactions')

# Display the first few rows
df.head()
```

```python
# Heat Map 1: Borrowing Transactions by Department and Year

# Load the Students sheet
students = pd.read_excel(
    'DAV_HEAT MAP GRAPHS.xlsx',
    sheet_name='Students'
)

# Check the column names
print("Students columns:")
print(students.columns.tolist())

# Merge transaction data with student information
merged_df = df.merge(
    students[['Student_ID', 'Department', 'Year']],
    on='Student_ID',
    how='left'
)

# Create Department × Year table
heatmap_data = pd.crosstab(
    merged_df['Department'],
    merged_df['Year']
)

print("Borrowing Transactions by Department and Year:")
display(heatmap_data)

# Create the heat map
plt.figure(figsize=(8, 5))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt='d',
    cmap='YlGnBu',
    linewidths=0.5
)

plt.title('Borrowing Transactions by Department and Year')
plt.xlabel('Year')
plt.ylabel('Department')

plt.tight_layout()
plt.show()
```

```python
# HEAT MAP 2 - Category vs Transaction Status

heatmap_data2 = pd.crosstab(
    df['Category'],
    df['Status']
)

plt.figure(figsize=(9, 5))

sns.heatmap(
    heatmap_data2,
    annot=True,
    fmt='d',
    cmap='YlOrRd',
    linewidths=0.5
)

plt.title('Borrowing Transactions by Category and Status')
plt.xlabel('Transaction Status')
plt.ylabel('Book Category')

plt.tight_layout()
plt.show()
```
